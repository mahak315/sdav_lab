```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

```

```python
marks=np.array([65,72,81,55,90])
print("Mean:",marks.mean())
print("Maximum:",marks.max())
print("Minimum:",marks.min())
print("Standard Deviation:",marks.std())
```

```python
data={
    "Name":["A","B","C","D","E"],
    "Marks":[65,72,81,55,90]
    }
df=pd.DataFrame(data)
print(df)
print(df["Marks"].mean())
```

```python
plt.bar(df["Name"],df["Marks"])
plt.xlabel("Student")
plt.ylabel("Marks")
plt.title("Student Marks")
plt.show()
```

```python
plt.hist(df["Marks"])
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.title("Marks Distribution")
plt.show()
```

```python
result=np.random.choice(["Head","Tail"],size=10)#choosing head or tail
print(result)
```

```python
result=np.random.choice(["Head","Tail"],size=10)
heads=np.sum(result=="Head")#count no of heads
print("Heads:",heads)
print(("Probability of head:",heads/10))#no of heads/10
```

```python
result=np.random.choice(["Head","Tail"],size=100)
heads=np.sum(result=="Head")#count no of heads
print("Heads:",heads)
print(("Probability of head:",heads/100))#no of heads/100
```

```python
result=np.random.choice(["Head","Tail"],size=1000)
heads=np.sum(result=="Head")#count no of heads
print("Heads:",heads)
print(("Probability of head:",heads/1000))#no of heads/1000
```

```python
counts=[np.sum(result=="Head"),np.sum(result=="Tail")]
plt.bar(["Head","Tail"],counts)
plt.xlabel("Outcome")
plt.ylabel("Count")
plt.title("Coin toss simulation")
plt.show()
```

```python
trials=[10,100,1000,10000]
for trial in trials:
    result=np.random.choice(["Head","Tail"],size=trial)
    probability=np.sum(result=="Head")/trial
    print(trial,probability)
```

```python
failed1st=0.25
failed_2nd=0.15
both=0.1
p1=both/failed_2nd #failed in 1st if already failed in 2nd
print(p1)
p2=both/failed1st #failed in 2nd if already failed in 1st
print(p2)
p3=failed1st+failed_2nd-both #failed in 1st or 2nd
print(p3)
```
